## 1. Library Installation and German IT Hub Dataset Creation
In this section, we install the Folium libray and construct a dataset containing the precise geographical coordinates (Latitude and Longitude) of major technology hubs in Germany, including München, Berlin, Frankfurt, and Hamburg.

```python
import pandas as pd
import folium

#creating a real-world dataset for German IT Hubs with coordinates

it_hub_data = {
    "City" : ["München", "Berlin","Frankfurt", "Hamburg"],
    "Region" :["Bavaria", "Berlin State", "Hesse", "Hamburg State"],
    "Latitude" : [48.1351, 52.2500, 50.1109, 53.5511],
    "Longitude" : [11.5820, 13.4050, 8.6821, 9.9937],
    "Major_Tech_Focus" : ["AI & Automotive Tech", "Startups & Fintech", "Data Centers & Cloud", "E-Commerce & Logistics"]
}

#Convert the dictionary into a Pandas DataFrame
df_hubs = pd.DataFrame(it_hub_data)

print("---GERMAN IT HUBS GEOSPATIAL DATASET---")
print(df_hubs)

#display data horizontally
print(df_hubs.T)
```

markdown
## 2. Generating the Interactive MAp and adding Geospatial Markers

In this section, we initialize a dynamic map centered on Germany use a loop to automatically drop custom markers on th coordinates of each IT hub.

```python

# 1. Initialize the map centered around Germany
# We use coordinates close to the center of Germany (Latitude 51.1657, Longitude 10.4515)
# zoom_start=6 is perfect to see the whole country at once

germany_map = folium.Map(location=[51.1657, 10.4515], zoom_start=6)

#2. Use a loop to iterate through the rows of our dataset and drop markers
for index, row in df_hubs.iterrows():
  #create an elegant information text for the popup box when clicked
  popup_text = f"City: {row['City']}<br>Focus: {row['Major_Tech_Focus']}"

  #Place a marker on the map using latitude and longitude
  folium.Marker(
      location=[row['Latitude'], row['Longitude']],
      popup=folium.Popup(popup_text, max_width=300),
      tooltip=row['City'], #shows the city name when mouse hovers over it
      icon=folium.Icon(color='purple', icon='info-sign')
  ).add_to(germany_map)

  #3. Display the interractive map directly in Google Colab
  germany_map

#nothing displayed because the code divided into codes, lets try again
```

```python
import pandas as pd
import folium

# 1. CREATE THE DATASET
it_hub_data = {
    "City": ["München", "Berlin", "Frankfurt", "Hamburg"],
    "Region": ["Bavaria", "Berlin State", "Hesse", "Hamburg State"],
    "Latitude": [48.1351, 52.5200, 50.1109, 53.5511],
    "Longitude": [11.5820, 13.4050, 8.6821, 9.9937],
    "Major_Tech_Focus": ["AI & Automotive Tech", "Startups & FinTech", "Data Centers & Cloud", "E-commerce & Logistics"]
}
df_hubs = pd.DataFrame(it_hub_data)

# 2. INITIALIZE THE GERMANY MAP WITH CARTO_DB TILES
# use "Cartodb Positron" for a clean and elegant light background
germany_map = folium.Map(location=[51.1657, 10.4515], zoom_start=6, tiles="Cartodb Positron")

# 3. LOOP TO DROP DYNAMIC PURPLE MARKERS
for index, row in df_hubs.iterrows():
    popup_text = f"City: {row['City']}<br>Focus: {row['Major_Tech_Focus']}"

    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=folium.Popup(popup_text, max_width=300),
        tooltip=row['City'],
        icon=folium.Icon(color='purple', icon='info-sign')
    ).add_to(germany_map)

# 4. DISPLAY THE INTERACTIVE MAP
germany_map

#the code running very well after 3 times trying
```
